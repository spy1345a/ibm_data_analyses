```python
import pandas as pd

# Student marks data
students = ['Alice', 'Bob', 'Charlie', 'David', 'Eva']
marks = [85, 78, 92, 88, 76]

# Create a Pandas Series
marks_series = pd.Series(marks, index=students)

# Display the series
print("Student Marks:")
print(marks_series)
```

```python
# Find the highest and lowest marks
highest_mark = marks_series.max()
lowest_mark = marks_series.min()
print("\nHighest Mark:", highest_mark)
print("Lowest Mark:", lowest_mark)
```

```python
# Find the average mark
average_mark = marks_series.mean()
print("Average Mark:", average_mark)
```

```python
# Find students who scored above 80
print("\nStudents scoring above 80:")
print(marks_series[marks_series > 80])
```
