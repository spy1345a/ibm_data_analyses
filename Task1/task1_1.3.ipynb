```python
import pandas as pd

# Grocery prices data
grocery_items = ['Milk', 'Bread', 'Eggs', 'Butter', 'Cheese']
prices = [2.5, 1.5, 3.0, 4.0, 5.0]

# Create a Pandas Series
price_series = pd.Series(prices, index=grocery_items)

# Display the series
print("Grocery Prices:")
print(price_series)
```

```python
# Find the most expensive item
most_expensive_item = price_series.idxmax()
highest_price = price_series.max()
print("Most Expensive Item:", most_expensive_item, "with price", highest_price)
```

```python
# Calculate the total cost of all items
total_cost = price_series.sum()
print("Total Cost of All Items:", total_cost)
```

```python
# Apply a discount of 10% on all items
discounted_prices = price_series * 0.9
print("\nPrices after 10% discount:")
print(discounted_prices)
```

```python
# Find items that cost more than $3
print("\nItems costing more than $3:")
print(price_series[price_series > 3])
```
