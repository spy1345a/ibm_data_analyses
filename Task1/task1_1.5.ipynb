```python
import pandas as pd

# Daily steps data
steps = [7000, 8500, 9000, 7500, 8000, 6500, 7000]
days = ['Mon', 'Tue', 'Wed', 'Thu', 'Fri', 'Sat', 'Sun']

# Create a Pandas Series
steps_series = pd.Series(steps, index=days)

# Display the series
print("Daily Steps:")
print(steps_series)
```

```python
# Find the total number of steps in the week
print("\nTotal steps in the week:", steps_series.sum())
```

```python
# Find the average number of steps per day
print("Average steps per day:", steps_series.mean())
```

```python
# Find the day with the maximum steps
print("Day with maximum steps:", steps_series.idxmax())
```

```python
# Check if you met the goal of 8000 steps each day
goal_met = (steps_series >= 8000).all()
print("Goal of 8000 steps met every day?:", goal_met)
```
