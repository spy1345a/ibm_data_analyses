```python
import pandas as pd

# Creating two Series with different indexes (months)
sales_2023 = pd.Series(
    [100, 200, 150],
    index=['Jan', 'Feb', 'Mar'],
    name='Sales_2023'
)

sales_2024 = pd.Series(
    [250, 300, 200],
    index=['Feb', 'Mar', 'Apr'],
    name='Sales_2024'
)

# Joining the Series into a DataFrame using an outer join
sales_df = pd.DataFrame(sales_2023).join(sales_2024, how='outer')

# Displaying the joined DataFrame
print("Joined Sales Data:")
print(sales_df)
```
