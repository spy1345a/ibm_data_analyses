```python
import pandas as pd

# Creating two separate Series
sales_product_A = pd.Series(
    [100, 200, 300],
    index=['Jan', 'Feb', 'Mar'],
    name='Product_A'
)

sales_product_B = pd.Series(
    [150, 250, 350],
    index=['Jan', 'Feb', 'Mar'],
    name='Product_B'
)

# Merging the Series into a DataFrame
sales_df = pd.DataFrame({
    'Product_A': sales_product_A,
    'Product_B': sales_product_B
})

# Displaying the merged DataFrame
print("Merged DataFrame:")
print(sales_df)
```
