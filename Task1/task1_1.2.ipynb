```python
import pandas as pd

# Daily temperature data
temperatures = [22, 24, 19, 23, 25, 21, 20]
days = ['Mon', 'Tue', 'Wed', 'Thu', 'Fri', 'Sat', 'Sun']

# Create a Pandas Series
temperature_series = pd.Series(temperatures, index=days)

# Display the series
print("Daily Temperatures:")
print(temperature_series)
```

```python
# Find the highest and lowest temperatures
highest_temp = temperature_series.max()
lowest_temp = temperature_series.min()

print("\nHighest Temperature:", highest_temp)
print("Lowest Temperature:", lowest_temp)
```

```python
# Calculate the temperature range
temp_range = highest_temp - lowest_temp
print("Temperature Range:", temp_range)
```

```python
# Check if any day had a temperature above 24 degrees
above_24 = (temperature_series > 24).any()
print("Any day above 24 degrees?:", above_24)
```
