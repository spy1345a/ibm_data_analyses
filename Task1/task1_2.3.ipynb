```python
import pandas as pd

# Creating two separate Series
population = pd.Series(
    [1000000, 500000, 1500000],
    index=['City_A', 'City_B', 'City_C'],
    name='Population'
)

area = pd.Series(
    [300, 200, 500],
    index=['City_A', 'City_B', 'City_C'],
    name='Area'
)

# Joining the two Series into a DataFrame
city_df = pd.DataFrame(population).join(area)

# Displaying the joined DataFrame
print("City Data:")
print(city_df)
```
