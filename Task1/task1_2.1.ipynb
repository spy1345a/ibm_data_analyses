```python
import pandas as pd

# Creating two separate Series
sales_product_A = pd.Series(
    [100, 200, 300],
    index=['Jan', 'Feb', 'Mar'],
    name='Product_A'
)

sales_product_B = pd.Series(
    [150, 250, 350],
    index=['Jan', 'Feb', 'Mar'],
    name='Product_B'
)
```

```python
# Displaying the original Series
print("Product A Sales:")
print(sales_product_A)

print("\nProduct B Sales:")
print(sales_product_B)
```

```python
# Concatenating the two Series
combined_sales = pd.concat([sales_product_A, sales_product_B])

# Displaying the concatenated Series
print("\nConcatenated Series:")
print(combined_sales)
```
