```python
import pandas as pd

# Monthly expenses data
expenses = [1200, 1500, 1000, 1800, 1300, 1600, 1100, 1400, 1900, 1700, 1250, 1450]
months = ['Jan', 'Feb', 'Mar', 'Apr', 'May', 'Jun', 'Jul', 'Aug', 'Sep', 'Oct', 'Nov', 'Dec']


# Create Pandas Series
expense_series = pd.Series(expenses, index=months)

print("Monthly Expenses:")
print(expense_series)
```

```python
# Total yearly expense
print("Total Expenses in a Year:", expense_series.sum())
```

```python
# Month with highest expense
print("Month with Highest Expense:", expense_series.idxmax(),"with",expense_series.max())
```

```python
# Average monthly expense
print("Average Monthly Expense:", expense_series.mean())
print("Average Monthly Expense(Rounded):",round(expense_series.mean()))
```

```python
# Expenses greater than 1500
print("Expenses greater than 1500:")
print(expense_series[expense_series > 1500])
```
