```python
import pandas as pd

# Creating two Series
revenue = pd.Series(
    [1000, 1500, 2000],
    index=['Jan', 'Feb', 'Mar'],
    name='Revenue'
)

units_sold = pd.Series(
    [50, 75, 100],
    index=['Jan', 'Feb', 'Mar'],
    name='Units_Sold'
)

# Merging the Series using join method
company_df = pd.DataFrame(revenue).join(units_sold)

# Displaying the merged DataFrame
print("Company Data:")
print(company_df)
```
